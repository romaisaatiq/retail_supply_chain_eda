# 5: Single Variable Analysis
Analyze individual columns using graphs and statistics.

# Sales Distribution

```python
# I used histogram to see how sales value are spread across the dataset
plt.figure(figsize=(9,5))
sns.histplot(df["Sales"],bins=40 ,kde=True, color="lightgreen")

plt.title("SALES DISTRIBUTION")
plt.xlabel("Sales")
plt.ylabel("Number of Records")

plt.xlim(0,2200)

plt.show()
```

# observation :
most order have lower sales,while a few order have higher sales

# Quantity Distribution

```python
#I uesd the bar/count plot to see which order quantity are most common

quantity_counts = df["Quantity"].value_counts().sort_index()
plt.figure(figsize=(9,5))
sns.barplot(x=quantity_counts.index,y=quantity_counts.values)

plt.title("NUMBER OF ORDERS BY QUANTITY")
plt.xlabel("Quantity")
plt.ylabel("Number of Order")


plt.show()

print("Most common quantity:",df["Quantity"].mode().iloc[0])
```

observation: The chart shows that some order quantities are more common than others,The most common quantity are shown by the highest bar 

# Number of Orders per Category

```python
# I used count plot to see the number of order in each category
plt.figure(figsize=(9,5))
sns.countplot(data = df, x= "Category", hue = "Category",palette = "viridis",legend = False )

plt.title("NUMBER OF ORDER BY CATEGORY")
plt.xlabel("Category")
plt.ylabel("Number of order")


plt.show()
```

Observation : This graph shows that furniture has highest number of order.while other has fewer order.

# Outlier detection in profit

```python
# I used box plot to detect the outlier in profit
plt.figure(figsize=(9,5))
sns.boxplot(data = df,x= "Profit",color= "plum")

plt.title("PROFIT DISTRIBUTION(OUTLIER DETECTION)")
plt.xlabel("Profit($)")

plt.xlim(-1000,1000)
plt.show()
```

Observation:

 The box plot shows that most profit values fall within a common range, while several unusually high and low values appear as outliers.

# 6: Relationship Analysis
Analyze relationship between two variable


# Sales and profit

```python
# I used sactter plot to see whether sales and profit tend to increase together
plt.figure(figsize=(9,5))

sns.scatterplot(data = df,x= "Sales",y= "Profit",hue= "Category")

plt.title("RELATIONSHIP BETWEEN SALES AND PROFIT")
plt.xlabel("Sales")
plt.ylabel("Profit")

plt.xlim(0,2000)
plt.ylim(500,-500)
plt.show()

```

 # Observation
 The scatter plot shows a generally positive relationship between Sales and Profit. However, some high-sales orders still generate low or negative profit, indicating that high sales do not always guarantee higher profitability.

# Total sales by region

```python
# I  used bar plot to compare total sales across different region
plt.figure(figsize=(9,5))

Region_sales = df.groupby("Region")["Sales"].sum().sort_values(ascending=False)
sns.barplot(x=Region_sales.index,y=Region_sales.values,hue=Region_sales.index, palette= "viridis" ,legend=False)

plt.title("TOTAL SALES BY REGION")
plt.xlabel("Region")
plt.ylabel("Sales ($)")

plt.show()


```

# observation :
 west region has highest sales value ,while south region has lowest sales value.

# 7: Monthly Sales Trend

```python
# I used the line plot to see how sales changed over time
plt.figure(figsize=(9,5))
df["Order Date"] = pd.to_datetime(df["Order Date"])
monthly_sales = df.groupby(df["Order Date"].dt.to_period("M"))["Sales"].sum()
monthly_sales.index=monthly_sales.index.to_timestamp()
plt.plot(monthly_sales.index,monthly_sales.values,marker="o")

plt.title("MONTHLY SALES TEND OVER TIME")
plt.xlabel("Months")
plt.ylabel("Total sales ($)")
plt.show()
```

# Observation:
Monthly sales fluctuate throughout the year, with stronger sales observed toward the end of the year, particularly during November and December.

# 8 Return Analysis

```python

# I used a count plot to compare returned and non-returned orders

plt.figure(figsize=(8,5))

sns.countplot(data=df, x="Returned")

plt.title("RETURNED VS NON-RETURNED ORDERS")
plt.xlabel("Returned")
plt.ylabel("Number of Orders")

plt.show()
```

# Observation :
The majority of orders were not returned, while only a smaller portion of orders were returned.

# 9: Correlation Analysis

```python
#I used the to find the relationship between numerical values
numeric_cols = df[["Sales","Quantity","Discount","Profit"]]
corr = numeric_cols.corr()

plt.figure(figsize=(9,5))

sns.heatmap(corr,annot=True,cmap="coolwarm")
plt.title("CORRELATION HEATMAP")
plt.show()
```

# Observation:
Sales and Profit show a positive relationship, while Discount and Profit show a negative relationship. Quantity has a relatively weak relationship with Profit.
